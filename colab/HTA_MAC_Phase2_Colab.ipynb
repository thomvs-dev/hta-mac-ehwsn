# HTA-MAC registered Phase 2 training on Google Colab

This notebook retrains the **18 preregistered Phase 2 models** on a Colab GPU:

- shared-branching budgets 8, 12, 16, 20, and 24;
- independent-DQN architecture ablation at budget 12;
- optimizer seeds 2299, 3299, and 4299;
- 500 episodes and 300 steps per episode;
- frozen development schedules 2300–2304, schema v2.

It does **not** retrain HEART-CH, change routing, tune on Phase 4 seeds, or alter the registered reward/configuration. Completed runs are copied to Google Drive after every model. If Colab disconnects, rerun the notebook: passed runs are restored and skipped.

Recommended runtime: **A100**, **L4**, or **T4 GPU** with high-RAM enabled when available. Do not use CPU unless there is no alternative.

```python
# Hardware/runtime and robust bundle setup.
import glob
import hashlib
import json
import os
import platform
import shutil
import stat
import sys
import zipfile
from pathlib import Path, PurePosixPath

import torch

print("Python:", sys.version)
print("Platform:", platform.platform())
print("Torch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

bundle_stem = "HTA_MAC_Colab_Training_Bundle_20260801"
expected_bundle_sha256 = (
    "b3ee26ed47d3a98ce03c232d7834d06e7b47bb7637fd44ea729ad6ccbaa44b79"
)
runtime_patterns = [
    f"/content/{bundle_stem}.zip",
    f"/content/{bundle_stem}*.zip",
]


def matching_zips(patterns):
    matches = []
    for pattern in patterns:
        matches.extend(Path(path) for path in glob.glob(pattern))
    return sorted({path.resolve() for path in matches if path.is_file()})


# Prefer a ZIP uploaded directly to the Colab runtime, including (1), (2), etc.
checked_patterns = list(runtime_patterns)
zip_candidates = matching_zips(runtime_patterns)
drive_module = None

# Search Google Drive only when no runtime upload is present.
if not zip_candidates:
    try:
        from google.colab import drive as drive_module

        drive_module.mount("/content/drive")
        drive_patterns = [
            f"/content/drive/MyDrive/{bundle_stem}.zip",
            f"/content/drive/MyDrive/{bundle_stem}*.zip",
        ]
        checked_patterns.extend(drive_patterns)
        zip_candidates = matching_zips(drive_patterns)
    except ImportError:
        local_patterns = [str(Path.cwd() / f"{bundle_stem}*.zip")]
        checked_patterns.extend(local_patterns)
        zip_candidates = matching_zips(local_patterns)

if not zip_candidates:
    raise FileNotFoundError(
        "Upload the HTA-MAC Colab ZIP before running this cell. Checked: "
        + str(checked_patterns)
    )

# Prefer the newest upload when Colab renamed duplicates.
bundle = max(zip_candidates, key=lambda path: path.stat().st_mtime_ns)
print("Using bundle:", bundle)

bundle_sha256 = hashlib.sha256(bundle.read_bytes()).hexdigest()
if bundle_sha256 != expected_bundle_sha256:
    raise RuntimeError(
        f"Wrong or incomplete bundle: SHA-256 {bundle_sha256}; "
        f"expected {expected_bundle_sha256}"
    )

workspace = Path("/content/stage2")
if workspace.exists():
    shutil.rmtree(workspace)

with zipfile.ZipFile(bundle, "r") as archive:
    bad_entry = archive.testzip()
    if bad_entry is not None:
        raise RuntimeError(f"Corrupt ZIP entry: {bad_entry}")

    normalized_names = [
        entry.filename.replace("\\", "/") for entry in archive.infolist()
    ]
    if len(normalized_names) != len(set(normalized_names)):
        raise RuntimeError("Archive contains colliding normalized paths")
    if "stage2/COLAB_BUNDLE_MANIFEST.json" not in set(normalized_names):
        raise RuntimeError(f"{bundle} is not the expected HTA-MAC Colab bundle")

    for entry, normalized_name in zip(archive.infolist(), normalized_names):
        normalized = PurePosixPath(normalized_name)
        parts = normalized.parts
        if normalized.is_absolute() or not parts or parts[0] != "stage2":
            raise RuntimeError(f"Unexpected archive entry: {entry.filename}")
        relative_parts = parts[1:]
        if not relative_parts:
            continue
        if any(part in ("", ".", "..") for part in relative_parts):
            raise RuntimeError(f"Unsafe archive entry: {entry.filename}")
        mode_type = (entry.external_attr >> 16) & 0o170000
        if mode_type == stat.S_IFLNK:
            raise RuntimeError(f"Symlink archive entry is not allowed: {entry.filename}")

        target = workspace.joinpath(*relative_parts)
        if entry.is_dir() or normalized_name.endswith("/"):
            target.mkdir(parents=True, exist_ok=True)
        else:
            target.parent.mkdir(parents=True, exist_ok=True)
            with archive.open(entry) as source, target.open("wb") as destination:
                shutil.copyfileobj(source, destination)

repo = workspace / "hta-mac"
upstream = workspace / "final_repo"
bundle_manifest_path = workspace / "COLAB_BUNDLE_MANIFEST.json"
if not repo.is_dir():
    raise RuntimeError(f"HTA-MAC repository is missing: {repo}")
if not upstream.is_dir():
    raise RuntimeError(f"Frozen HEART-CH repository is missing: {upstream}")
if not bundle_manifest_path.is_file():
    raise RuntimeError(f"Bundle manifest is missing: {bundle_manifest_path}")

# Verify every file recorded by the bundle builder.
bundle_manifest = json.loads(bundle_manifest_path.read_text(encoding="utf-8-sig"))
manifest_failures = []
for record in bundle_manifest["files"]:
    relative = PurePosixPath(record["path"])
    if relative.is_absolute() or any(part in ("", ".", "..") for part in relative.parts):
        manifest_failures.append(f"unsafe:{record['path']}")
        continue
    target = workspace.joinpath(*relative.parts)
    if not target.is_file():
        manifest_failures.append(f"missing:{record['path']}")
        continue
    if target.stat().st_size != int(record["bytes"]):
        manifest_failures.append(f"size:{record['path']}")
        continue
    actual = hashlib.sha256(target.read_bytes()).hexdigest()
    if actual != record["sha256"]:
        manifest_failures.append(f"sha256:{record['path']}")

if manifest_failures:
    raise RuntimeError(
        f"Bundle manifest verification failed ({len(manifest_failures)}): "
        + str(manifest_failures[:10])
    )

# Drive is required for per-model backups even when the ZIP came from /content.
if drive_module is None:
    from google.colab import drive as drive_module

    drive_module.mount("/content/drive")

os.chdir(repo)
for import_path in (str(repo), str(upstream)):
    if import_path not in sys.path:
        sys.path.insert(0, import_path)

print("BUNDLE_SHA256_PASS=True")
print(f"BUNDLE_MANIFEST_FILES={len(bundle_manifest['files'])}")
print("BUNDLE_MANIFEST_FAILURES=0")
print("Repository:", repo)
print("Frozen HEART-CH:", upstream)
print("HTA-MAC source commit:", bundle_manifest["hta_mac_git_commit"])
print("Frozen upstream commit:", bundle_manifest["frozen_heart_ch_git_commit"])
```

```python
import json
import shutil
import subprocess
import sys
import time
from datetime import datetime, timezone
from pathlib import Path

from experiments.run_phase2_registered_sweep import (
    command_for,
    read_existing,
    registered_runs,
)

def parse_selection(text, upper):
    selected = set()
    for token in text.split(","):
        token = token.strip()
        if not token:
            continue
        if "-" in token:
            start, stop = map(int, token.split("-", 1))
            selected.update(range(start, stop + 1))
        else:
            selected.add(int(token))
    if not selected or min(selected) < 0 or max(selected) >= upper:
        raise ValueError(f"RUN_SELECTION must use indices 0 through {upper - 1}")
    return selected

all_runs = registered_runs()
selected = parse_selection(RUN_SELECTION, len(all_runs))
drive_root = Path(DRIVE_OUTPUT_DIR)
drive_runs = drive_root / "runs"
drive_runs.mkdir(parents=True, exist_ok=True)
local_phase2 = repo / "outputs" / "phase2"
local_phase2.mkdir(parents=True, exist_ok=True)

# Restore every completed Drive run first. This makes repeated Colab sessions resumable.
for run in all_runs:
    source = drive_runs / run["run_name"]
    target = local_phase2 / run["run_name"]
    if source.is_dir():
        shutil.copytree(source, target, dirs_exist_ok=True)

def gate_pass(run):
    try:
        summary = read_existing(run)
    except Exception as exc:
        print(f"Existing run rejected: {run['run_name']}: {exc}")
        return False
    return bool(summary and summary.get("phase2_curriculum_gate_pass"))

def write_registry():
    rows = []
    for index, run in enumerate(all_runs):
        row = dict(run)
        row["index"] = index
        row["status"] = "gate_pass" if gate_pass(run) else "pending"
        row["summary"] = str(
            Path("outputs") / "phase2" / run["run_name"] / "summary.json"
        )
        rows.append(row)
    passed = sum(row["status"] == "gate_pass" for row in rows)
    registry = {
        "created_utc": datetime.now(timezone.utc).isoformat(),
        "execution_platform": "google_colab",
        "source_git_commit": subprocess.check_output(
            ["git", "rev-parse", "HEAD"], cwd=repo, text=True
        ).strip(),
        "preregistration": "PHASE4_PREREGISTRATION.md",
        "schedule_schema_version": 2,
        "episodes": 500,
        "max_steps": 300,
        "device": device,
        "gpu": gpu_name,
        "runs": rows,
        "expected_runs": len(rows),
        "gate_pass_runs": passed,
        "complete": passed == len(rows),
    }
    path = local_phase2 / "registered_sweep_registry.json"
    path.write_text(json.dumps(registry, indent=2), encoding="utf-8")
    shutil.copy2(path, drive_root / path.name)
    return registry

for index, run in enumerate(all_runs):
    if index not in selected:
        continue
    if gate_pass(run):
        print(f"SKIP_GATE_PASS index={index} run={run['run_name']}", flush=True)
        continue

    print(
        f"START index={index}/17 architecture={run['architecture']} "
        f"budget={run['budget']} seed={run['optimizer_seed']}",
        flush=True,
    )
    started = time.time()
    command = command_for(run, episodes=500, max_steps=300, device=device)
    completed = subprocess.run(command, cwd=repo, check=False)
    elapsed = time.time() - started
    run_dir = local_phase2 / run["run_name"]
    runtime_record = {
        "platform": "google_colab",
        "gpu": gpu_name,
        "torch_version": torch.__version__,
        "cuda_version": torch.version.cuda,
        "elapsed_seconds": elapsed,
        "return_code": completed.returncode,
        "command": command,
        "completed_utc": datetime.now(timezone.utc).isoformat(),
    }
    run_dir.mkdir(parents=True, exist_ok=True)
    (run_dir / "colab_runtime.json").write_text(
        json.dumps(runtime_record, indent=2), encoding="utf-8"
    )

    if not gate_pass(run):
        shutil.copytree(
            run_dir,
            drive_runs / run["run_name"],
            dirs_exist_ok=True,
        )
        write_registry()
        raise RuntimeError(
            f"Registered gate failed for {run['run_name']}. "
            "Training stopped; do not continue with a broken run."
        )

    shutil.copytree(
        run_dir,
        drive_runs / run["run_name"],
        dirs_exist_ok=True,
    )
    registry = write_registry()
    print(
        f"BACKUP_COMPLETE run={run['run_name']} "
        f"gate_pass={registry['gate_pass_runs']}/18 "
        f"elapsed_minutes={elapsed / 60:.1f}",
        flush=True,
    )

registry = write_registry()
print(json.dumps({
    "selected_indices": sorted(selected),
    "gate_pass_runs": registry["gate_pass_runs"],
    "expected_runs": registry["expected_runs"],
    "complete": registry["complete"],
}, indent=2))
```

```python
import json
import shutil
import subprocess
import sys
from pathlib import Path

registry_path = repo / "outputs" / "phase2" / "registered_sweep_registry.json"
registry = json.loads(registry_path.read_text(encoding="utf-8"))
if not registry["complete"]:
    print(
        f"Phase 2 is not complete yet: {registry['gate_pass_runs']}/18 gates passed. "
        "Rerun with the missing indices; completed Drive runs will be restored."
    )
else:
    subprocess.run(
        [sys.executable, "-B", "experiments/archive_registered_phase2.py"],
        cwd=repo,
        check=True,
    )
    manifest_path = (
        repo / "outputs" / "phase2" / "registered_sweep_artifact_manifest.json"
    )
    shutil.copy2(manifest_path, Path(DRIVE_OUTPUT_DIR) / manifest_path.name)
    shutil.copy2(registry_path, Path(DRIVE_OUTPUT_DIR) / registry_path.name)

    evidence = {
        "registry": json.loads(registry_path.read_text(encoding="utf-8")),
        "artifact_manifest_sha256": sha256_file(manifest_path),
        "artifact_manifest": str(manifest_path),
    }
    evidence_path = Path(DRIVE_OUTPUT_DIR) / "COLAB_PHASE2_COMPLETION.json"
    evidence_path.write_text(json.dumps(evidence, indent=2), encoding="utf-8")
    print("REGISTERED_PHASE2_COMPLETE=True")
    print("Drive evidence:", evidence_path)
```
